# Exploratory data analysis of the HR/LR dataset

Measures how far each LR image is from its HR reference and writes the
figures under `results/eda`. The logic lives in `srlib/eda.py`.

Every LR image is brought up to the HR frame with a single declared
interpolation, the same one SRCNN is fed. The per-image degradation kernel
recorded in the degradation log is deliberately not used: an LR capture in
the wild does not come with the filter that produced it, and measuring each
image under a different kernel would make them incomparable with each other.

```python
import os
import sys

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "../")))

from srlib.eda import EDAPipeline
```

## Run the analysis

```python
# Paths default to the dataset roots declared in constants.
pipeline = EDAPipeline()

df = pipeline.run()
df
```
